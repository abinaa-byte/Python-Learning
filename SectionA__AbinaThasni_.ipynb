```python
armstrong_numbers = []

for number in range(1, 10000):
    temp = number
    digits = 0

    while temp > 0:
        digits += 1
        temp = temp // 10

    temp = number
    total = 0

    while temp > 0:
        digit = temp % 10
        total = total + (digit ** digits)
        temp = temp // 10

    if total == number:
        armstrong_numbers.append(number)

print(", ".join(map(str, armstrong_numbers)))
print("Count:", len(armstrong_numbers))
```

```python
n = int(input("Enter N: "))

is_prime = True

if n < 2:
    is_prime = False
else:
    i = 2

    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1

if is_prime:
    print(n, "is prime")
else:
    print(n, "is not prime")

total = 0

for number in range(2, n + 1):
    prime = True
    i = 2

    while i * i <= number:
        if number % i == 0:
            prime = False
            break
        i += 1

    if prime:
        total = total + number

print("SUM of primes:", total)
```

```python
marks = list(map(int, input("Enter 10 marks: ").split()))

maximum = marks[0]
minimum = marks[0]
total = 0

for mark in marks:
    if mark > maximum:
        maximum = mark

    if mark < minimum:
        minimum = mark

    total = total + mark

average = total / 10

above_average = 0

for mark in marks:
    if mark > average:
        above_average += 1

highest = marks[0]
second_highest = marks[0]

for mark in marks:
    if mark > highest:
        second_highest = highest
        highest = mark
    elif mark > second_highest and mark != highest:
        second_highest = mark

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Average:", average)
print("Students above average:", above_average)
print("Second highest:", second_highest)
```

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)

days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

maximum = max(prices)
minimum = min(prices)

max_day = days[prices.index(maximum)]
min_day = days[prices.index(minimum)]

average = sum(prices) / len(prices)

biggest_gain = 0
gain_day = ""

volatility = 0

for i in range(1, len(prices)):
    change = prices[i] - prices[i - 1]

    if change > biggest_gain:
        biggest_gain = change
        gain_day = days[i - 1] + " to " + days[i]

    volatility = volatility + abs(change)

print("Maximum price:", maximum, "on", max_day)
print("Minimum price:", minimum, "on", min_day)
print("Average price:", round(average, 2))
print("Biggest single-day gain:", biggest_gain, "between", gain_day)
print("Total volatility:", volatility)
```

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive == False:
        base_amount = amount
        gst_amount = amount * gst_rate / 100
        total_amount = amount + gst_amount

    else:
        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount

    return (round(base_amount, 2), round(gst_amount, 2), round(total_amount, 2))


print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

Q47

```python
def power(base, exp):
    if exp == 0:
        return 1

    if exp < 0:
        return 1 / power(base, -exp)

    return base * power(base, exp - 1)


print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

# (a) Total weekly sales
total_sales = np.sum(sales)

# (b) Daily average
average_sales = np.mean(sales)

# (c) Day with maximum sales
maximum_sales = np.max(sales)
maximum_day = days[np.argmax(sales)]

# (d) Number of days above the weekly average
days_above_average = np.sum(sales > average_sales)

# (e) Percentage growth from Monday to Sunday
growth = ((sales[-1] - sales[0]) / sales[0]) * 100

# (f) Day-on-day percentage changes
daily_changes = ((sales[1:] - sales[:-1]) / sales[:-1]) * 100

print("Total weekly sales:", total_sales)
print("Daily average:", round(average_sales, 2))
print("Maximum sales:", maximum_sales, "on", maximum_day)
print("Days above average:", days_above_average)
print("Monday to Sunday growth:", round(growth, 2), "%")
print("Day-on-day % changes:", np.round(daily_changes, 2))
```
